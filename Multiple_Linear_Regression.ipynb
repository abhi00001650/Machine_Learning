```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_diabetes
```

```python
diabetes = load_diabetes()
```

```python
data = pd.DataFrame(diabetes.data , columns = diabetes.feature_names)
```

```python
data
```

```python
data['target'] = diabetes.target
```

```python
data
```

```python
#EDA and data preparation

X = data.drop('target' , axis = 1)
Y = data['target']
```

```python
X
```

```python
Y
```

```python
#split data in train nad test
from sklearn.model_selection import train_test_split
X_train , X_test , Y_train , Y_test = train_test_split(X,Y, test_size = 0.2 , random_state = 1)
```

```python
X_train.shape , X_test.shape 
```

```python
#model training 

from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
model.fit(X_train , Y_train) 
```

```python
X_train.shape
```

```python
model.coef_
```

```python
len(model.coef_)
```

```python
model.intercept_
```

```python
#interpretation
#say age has coef value -30.62219046
#interpretation> 1 unit change in age then target value decreases by -30 units on an average when all
#other features/factors are kept constant
X_train
```

```python
Y_pred = model.predict(X_test)
Y_pred
```

```python
from sklearn.metrics import mean_squared_error , r2_score  
```

```python
mean_squared_error(Y_test , Y_pred)
```
