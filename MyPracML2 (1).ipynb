```python
"""
* Imbalanced Data -> Supervised Learning -> Regression (y is continuous)
                                                     and
                                            Classification (y is discrete)
                                            
               y= (pass/fail) OR (diabetic/non-diabetic)

*2 outcomes of target variable (y)
     -> binary classification problem
       (bi)-> means two

  Sugar level      cholestrol          y(diabetic/non-diabetic)
     (f1)              (f2)                   
     250               102                   1 ->(diabetic)
     300               100                   1
     400               200                   0 -> (non-diabetic)


    -> 90% - 0 (non-diabetic)
       10% - 1 (diabetic)

    * Assuming you dont build any model and you predict everything as class 0. It will
      be having 90% of accuracy as 90% of data is non diabetic.

      -> class imbalance -> when 1 class has very high percentage as compared to other 
         class this is class imbalance.
         90% class 0 -> majority class(non-diabetic)
         10% class 1 -> minority class(diabetic)

 *you dont want your model to learn only the data pattern which is in majority ->
 -> you need to deal with class imbalance

 Class Imbalance -> when the diff in the count/percentage of both the class is huge.

                    Class Imbalance 80% - class 0
                                    20% - class 1

* Solution to class imbalance->
1) Undersampling(down sampling)
2) Oversampling(up sampling)
3) SMOTE


1) Undersampling (Downsample) -> majority class is downsampled . 
disadvantage - ML is about learning pattern in data
               

2) Oversampling (Upsample) -> minority class is upsampled . Repeat the minority class data.
disadvantage - Data is just repeating in case of over sampling
               No Pattern
               Noise

3) SMOTE (Synthetic Minority Oversampling Technique) -> Problem of no pattern is solved as seen 
                                                        in oversampling.
                                                     -> Like datapoint will always be together thatswhy 
                                                        SMOTE is effective as it is creating new datapoints

Q) How to use in industry -> We can use combination of SMOTE AND Undersampling 
                            Dec. 900 to 600
                            Inc. 100 to 400
                            It is a good ratio 60:40
 * In practical impletmentation of class imbalance .
     population                   with replacement     without replacement
     [1,2,3,4,5]                    S1(1,2)            S1(1,2)
                                    S2(1,3)            S2(3,4)
                                    S3(2,3)            S3(5,6)
                                    data point         data point
                                    can be             cannot be
                                    repeated           repeated
"""
```

```python
#PRACTICAL IMPLEMENTATION

#class imbalance > when 1 class has higher percentage
#handle
#upsampling
#downsampling
#SMOTE

import pandas as pd
import numpy as np

np.random.seed(1) #for reproducibility of result

no_samples = 1000
class_0_ratio = 0.9
no_class_0 =int(no_samples*class_0_ratio)
no_class_1 = 100
```

```python
no_class_0,no_class_1
```

```python
class_0 = {'feature1' : np.random.normal(0,1,no_class_0),
           'feature2' : np.random.normal(0,1,no_class_0),
           'target' : [0]*no_class_0}
```

```python
class_0
```

```python
class_0 = pd.DataFrame(class_0)
```

```python
class_0
```

```python
class_1 = pd.DataFrame({'feature1' : np.random.normal(3,1,no_class_1),
           'feature2' : np.random.normal(3,1,no_class_1),
           'target' : [1]*no_class_1})
```

```python
class_1
```

```python
df = pd.concat([class_0,class_1]).reset_index(drop = True)
```

```python
df
```

```python
df.target.value_counts()
```

```python
#upsampling 

df_minority = df[df.target == 1]
df_majority = df[df.target == 0]
```

```python
df_majority
```

```python
df_minority
```

```python
 #replace >> if replace is true, it simply means datapoint can be repeated
from sklearn.utils import resample
df_minority_upsampled = resample(df_minority , replace = True, n_samples = len(df_majority), random_state = 1)
```

```python
df_minority_upsampled.shape
```

```python
df_upsampled = pd.concat([df_majority,df_minority_upsampled])
```

```python
df_upsampled.target.value_counts()
```

```python
#downsampling 
df_minority
```

```python
df_majority
```

```python
df_majority_downsampled = resample(df_majority,replace = False , n_samples = len(df_minority) , random_state = 1)
```

```python
df_majority_downsampled
```

```python
df_downsampled = pd.concat([df_minority , df_majority_downsampled])
```

```python
import numpy as np
import matplotlib.pyplot as plt
#linear interpolation
x = np.array([1,2,3,4,5])
y = np.array([1,3,5,7,9])

plt.scatter(x,y)
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings("ignore")
```

```python
salary = [11, 40, 45, 68, 65, 68, 78, 90, 57, 74, 91, 92, 88, 68, 57, 48, 99, 101, 68, 77, 110, 140]
salary
```

```python
#5 point summary 
#min value
#25th
#50th
#75th
#100th
```

```python
np.quantile(salary , [0 , 0.25, 0.50, 0.75, 1.00])
```

```python
df = pd.DataFrame(salary, columns = ['Salary'])
```

```python
df.describe()
```

```python
#To check outliers>> distplot , boxplot

plt.figure(figsize = (12,5))
plt.subplot(1,2,1)
sns.histplot(df['Salary'] , kde = True)
plt.title("Dist Plot")


plt.subplot(1,2,2)
sns.boxplot(data = df, x = 'Salary')
plt.title("Box plot")
plt.show()
```

```python
#here the distribution is not skewed as the outliers are present on both side
#box plot we see there are outliers 

#1)DROP THE OUTLIERS
#2)CAPPING THE OUTLIERS
#3)REPLACING WITH MEAN AND MEDIAN
```

```python
#1)DROPPING THE OUTLIERS
```

```python
Q1 = df['Salary'].quantile(0.25)
Q3 = df['Salary'].quantile(0.75)

IQR = Q3-Q1

lower_fence = Q1-1.5*IQR
upper_fence = Q3+1.5*IQR
```

```python
df_filtered = df[(df.Salary >= lower_fence) & (df.Salary <= upper_fence)]
```

```python
df_filtered.shape
```

```python
plt.figure(figsize = (12,5))
plt.subplot(1,2,1)
sns.histplot(df_filtered['Salary'] , kde = True)
plt.title("Dist Plot")


plt.subplot(1,2,2)
sns.boxplot(data = df_filtered, x = 'Salary')
plt.title("Box plot")
plt.show()
```

```python
#imputation with mean and median
df['Salary_imputed_mean'] = np.where((df.Salary >= upper_fence) | (df.Salary <= lower_fence), df['Salary'].mean(), df['Salary'])
```

```python
df
```

```python
plt.figure(figsize = (12,5))
plt.subplot(1,2,1)
sns.histplot(df['Salary_imputed_mean'] , kde = True)
plt.title("Dist Plot")


plt.subplot(1,2,2)
sns.boxplot(data = df, x = 'Salary_imputed_mean')
plt.title("Box plot")
plt.show()
```

```python
#for median
df['Salary_imputed_median'] = np.where((df.Salary >= upper_fence) | (df.Salary <= lower_fence), df['Salary'].median(), df['Salary'])
```

```python
df
```

```python
plt.figure(figsize = (12,5))
plt.subplot(1,2,1)
sns.histplot(df['Salary_imputed_median'] , kde = True)
plt.title("Dist Plot")


plt.subplot(1,2,2)
sns.boxplot(data = df, x = 'Salary_imputed_median')
plt.title("Box plot")
plt.show()
```

```python
#CAPPING THE OUTLIERS
#CAPPING>> replacing the nearest value which is not outlier
df
```

```python
lower_cap = df['Salary'].quantile(0.05) # lower cap is the 5th percentile
upper_cap = df['Salary'].quantile(0.95) #upper cap as 95th percentile~
```

```python
lower_cap
```

```python
upper_cap
```

```python
#replace outliers with cap

df['Salary_capped'] = np.where(df['Salary'] < lower_cap , lower_cap, 
        np.where(df['Salary'] > upper_cap , upper_cap, df['Salary']))
```

```python
df
```

```python
plt.figure(figsize = (12,5))
plt.subplot(1,2,1)
sns.histplot(df['Salary_capped'] , kde = True)
plt.title("Dist Plot")


plt.subplot(1,2,2)
sns.boxplot(data = df, x = 'Salary_imputed_median')
plt.title("Box plot")
plt.show()
```
