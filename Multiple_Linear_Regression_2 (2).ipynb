```python
#POLYNOMIAL REGRESSION

import numpy as np
import matplotlib.pyplot as plt
```

```python
#NP.RANDOM same synthetich dp or reproducibilty of results
np.random.seed(1)
```

```python
#Genreate synthetic X
X = 2*np.random.rand(100 , 1 )
```

```python
X
```

```python
#Generate random synthtic Y
Y = 4 + 3 * X + 1.5 * X ** 2 + np.random.randn(100 , 1)
```

```python
Y
```

```python
#split data in train nad test
from sklearn.model_selection import train_test_split
X_train , X_test , Y_train , Y_test = train_test_split(X,Y, test_size = 0.2 , random_state = 1)
```

```python
X_train.shape , X_test.shape ,Y_train.shape , Y_test.shape 
```

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
poly_features = PolynomialFeatures(degree = 2 , include_bias = False)
X_poly_train = poly_features.fit_transform(X_train)
```

```python
X_poly_train
```

```python
from sklearn.linear_model import LinearRegression
poly_reg = LinearRegression()
```

```python
poly_reg
```

```python
poly_reg.fit(X_poly_train , Y_train)
```

```python
poly_reg.coef_
```

```python
poly_reg.intercept_
```

```python
#Make Predcition for train data itself for the evaluation matrix
```

```python
Y_poly_predict = poly_reg.predict(X_poly_train)
```

```python
Y_poly_predict
```

```python
from sklearn.metrics import mean_squared_error
mean_squared_error(Y_train , Y_poly_predict)
```

```python
plt.scatter(X_train , Y_train , label = 'traindata') 
```

```python
degree = 2
plt.scatter(X_train, Y_train, label = "training data")
X_range = np.linspace(0, 2, 100).reshape(-1, 1)
X_range_poly = poly_features.transform(X_range)
plt.plot(X_range, poly_reg.predict(X_range_poly), color='red', label=f'Polynomial Regression (Degree {degree})')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Polynomial Regression')
plt.legend()
plt.show()
```
