```python
data = pd.DataFrame(diabetes.data , columns = diabetes.feature_names)
```

```python
data.head()
```

```python
data['target'] = diabetes.target
```

```python
data
```

```python
#EDA ,Dtacleaning , DataPrep , Feature Engineering
```

```python
X = data[['bmi']] # 1 independent variable
Y = data['target'] # dependent variable
```

```python
X
```

```python
Y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, Y_train, Y_test = train_test_split( X , Y , test_size = 0.2 , random_state = 1)
```

```python
#scaling (optional)
```

```python
#Model Training

from sklearn.linear_model import LinearRegression
```

```python

model = LinearRegression()
```

```python
model
```

```python
X_train
```

```python
Y_train
```

```python
model.fit(X_train , Y_train) #training of model
```

```python
model.coef_ # attribiue of model
```

```python
model.intercept_
```

```python
Y_pred = model.predict(X_test)
```

```python
Y_pred
```

```python
#Visualise the result
plt.scatter(X_test , Y_test , color ='black' ,  label = 'Actual Data')
plt.plot(X_test , Y_pred , color = 'blue' , linewidth = 3 , label = ' Predicted Line')
plt.xlabel('BMI')
plt.ylabel('1 year progression Targte')
plt.title('Linear regression on diabetes data')
plt.legend()
plt.plot()
```

```python
#model evaluaton

from sklearn.metrics import mean_squared_error
mean_squared_error(Y_test , Y_pred)
```
