ASSIGNMENT QUESTIONS

1) What is a parameter?
-  A parameter is a value learned by a model during training that helps make predictions.

2. What is correlation?
-  Correlation measures the relationship between two variables.

3. What does negative correlation mean?
-  When one variable increases, the other decreases.

4. Define Machine Learning. What are the main components?
-  Machine Learning is a method where machines learn patterns from data.
Main components: Data, Features, Model, Algorithm, Loss function, Optimizer.

5. How does loss value help determine model quality?
-  Lower loss means better model performance.

6. What are continuous and categorical variables?
-  Continuous: Numerical values ( height, weight).
Categorical: Labels or categories ( color, gender).

7. How do we handle categorical variables? Common techniques?
-  By encoding them. Techniques: Label Encoding, One-Hot Encoding.

8. What do you mean by training and testing a dataset?
-  Training data trains the model; testing data checks model performance.

9. What is sklearn.preprocessing?
-  A module in scikit-learn used for data scaling and encoding.

10. What is a Test set?
-  A dataset used to evaluate the trained model.

11. How do we split data in Python?
-  Using train_test_split() from sklearn.model_selection.

12. How do you approach a Machine Learning problem?
-  Understand problem -> Collect data -> EDA -> Preprocessing -> Train model -> Evaluate -> Improve.

13. Why perform EDA before modeling?
-  To understand data, detect patterns, missing values, and outliers.

14. What is correlation?
-  Relationship between variables.

15. What does negative correlation mean?
-  One increases while the other decreases.

16. How to find correlation in Python?
-  Using df.corr() or numpy.corrcoef().

17. What is causation? Difference between correlation & causation?
-  Causation means one variable causes another.
Correlation shows relation, not cause.
Example: Ice cream sales and drowning are correlated, but ice cream does not cause drowning.

18. What is an Optimizer? Types with examples?
-  Optimizer minimizes loss.
Types: SGD, Adam, RMSprop.
Example: Adam is widely used for fast convergence.

19. What is sklearn.linear_model?
-  A module containing linear models like Linear Regression and Logistic Regression.

20. What does model.fit() do? Arguments?
-  Trains the model. Arguments: X_train, y_train.

21. What does model.predict() do? Arguments?
-  Makes predictions. Argument: X_test.

22. What are continuous and categorical variables?
-  Continuous: numeric values
Categorical: category labels

23. What is feature scaling? How does it help?
-  Rescaling values to a common range; improves model performance and speed.

24. How do we perform scaling in Python?
-  Using StandardScaler or MinMaxScaler.

25. What is sklearn.preprocessing?
-  Module for scaling, encoding, and normalization.

26. How do we split data in Python?
-  Using train_test_split().

27. Explain data encoding.
-  Converting categorical data into numeric format using methods like Label Encoding and One-Hot Encoding.
