```python
import plotly.express as px

# Group by Position for terminated employees
terminated_counts = df[df['Termd'] == 1].groupby('Position')['Employee_Name'].count().reset_index()
terminated_counts.rename(columns={'Employee_Name':'Terminated_Count'}, inplace=True)

# Create bar chart
fig = px.bar(
    terminated_counts,
    x='Terminated_Count',
    y='Position',
    orientation='h',  # Horizontal bars
    color='Terminated_Count',
    color_continuous_scale='Blues',
    title='Number of Terminated Employees by Position'
)

fig.update_layout(
    xaxis_title='Number of Terminated Employees',
    yaxis_title='Position',
    yaxis={'categoryorder':'total ascending'}  # Sort bars by count
)

fig.show()
```
